```python
import pandas as pd

# Actual path to the Tamil sentiment dataset
train_path = '/content/DravidianCodeMix-Dataset/programs/tamil_sentiment_full.csv'

# Read the CSV file into a DataFrame, explicitly telling pandas there's no header.
# Since `sep='\t'` was not effective, we'll read it as a single column and split manually.
df = pd.read_csv(train_path, header=None, engine='python', on_bad_lines='warn')

# The DataFrame `df` will now have a single column (named 0). This column contains
# strings like 'label\ttext'. We need to split this column into two.
# Use n=1 to ensure it splits only on the first tab, in case the text itself contains tabs.
df_split = df.iloc[:, 0].str.split('\t', n=1, expand=True)

# Assign the split parts to new 'label' and 'text' columns.
# Also, drop any original columns and keep only the new 'label' and 'text' columns.
if df_split.shape[1] == 2:
    df = df_split.rename(columns={0: 'label', 1: 'text'})
else:
    print("Warning: The file could not be reliably split into 'label' and 'text' columns.")
    # Fallback to assign what we can, which might still cause issues downstream if not properly split
    df['label'] = df.iloc[:,0].apply(lambda x: x.split('\t',1)[0] if '\t' in x else 'unknown_label')
    df['text'] = df.iloc[:,0].apply(lambda x: x.split('\t',1)[1] if '\t' in x else x)
    df = df[['label', 'text']] # Ensure only these two columns are present

df.head(10)
```

```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Make sure the folder exists first (this was likely the actual missing step)
os.makedirs('/content/drive/MyDrive/mozhi', exist_ok=True)

# 2. Load raw data
repo_path = '/content/DravidianCodeMix-Dataset'
if not os.path.exists(repo_path):
    !git clone https://github.com/bharathichezhiyan/DravidianCodeMix-Dataset.git

train_path = os.path.join(repo_path, 'programs/tamil_sentiment_full.csv')
df = pd.read_csv(train_path, header=None, engine='python', on_bad_lines='warn')

df_split = df.iloc[:, 0].str.split('\t', n=1, expand=True)
df = df_split.rename(columns={0: 'label', 1: 'text'})

# 3. Clean labels
df['label'] = df['label'].str.strip()
label_map = {
    'Positive': 'positive',
    'Negative': 'negative',
    'unknown_state': 'neutral'
}
df_clean = df[df['label'].isin(label_map.keys())].copy()
df_clean['label'] = df_clean['label'].map(label_map)

# 4. Split
train_df, temp_df = train_test_split(
    df_clean, test_size=0.2, stratify=df_clean['label'], random_state=42
)
val_df, test_df = train_test_split(
    temp_df, test_size=0.5, stratify=temp_df['label'], random_state=42
)

# 5. Save
train_df.to_csv('/content/drive/MyDrive/mozhi/train.csv', index=False)
val_df.to_csv('/content/drive/MyDrive/mozhi/val.csv', index=False)
test_df.to_csv('/content/drive/MyDrive/mozhi/test.csv', index=False)

# 6. CONFIRM it actually worked before trusting it
print("Files in mozhi folder:", os.listdir('/content/drive/MyDrive/mozhi'))
print("Train rows:", len(train_df), "| Val rows:", len(val_df), "| Test rows:", len(test_df))
```
